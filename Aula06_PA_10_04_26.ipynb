```python
# Estrutura de Decisão

numero = 5
print ("-=-=-Estrutura de Decisão-=-=-")
print (" ")
numero01 = int (input ("Digite um Número:"))

print ("-=-"*10)

if numero > numero01:
  print (f"{numero} é maior que {numero01}")

elif numero < numero01:
   print (f"{numero} é menor que {numero01}")

else:
    print (f"{numero} é igual a {numero01}")
```

```python
# Escrever um programa que leia dois números e retorne se eles são iguais

print ("-=-=-=-Leitor de Números-=-=-=-")
print (" ")

a = int (input ("Digite um número:"))
b = int (input ("Digite um número:"))

print ("-=-"*10)

if a > b :
  print (f"{a} maior que {b}")

else:
    print (f"{a} menor que {b}")
```

```python
# Escrever média anual
# Média menor que 5 - Reprovado
# Média menor que 7, maior que 5 - Recuperação
# Média maior que 7 - Aprovado

print ("=-=-=-=-=-MÉDIA ANUAL-=-=-=-=-=")
print (" ")

nome = str (input ("Digite o nome:"))
print ("-=-"*10)

n1 = float (input ("Digite a nota do primeiro bimestre:"))
n2 = float (input ("Digite a nota do segundo bimestre:"))
n3 = float (input ("Digite a nota do terceiro bimestre:"))
n4 = float (input ("Digite a nota do quarto bimestre:"))

print ("-=-"*10)

media = (n1 + n2 + n3 + n4) / 4

if media >= 7:
  print (f"{nome} foi Aprovado!")

elif 5 <= media > 7:
  print (f"{nome} está de Recuperação!")

else:
  print (f"{nome} foi Reprovado!")
```
