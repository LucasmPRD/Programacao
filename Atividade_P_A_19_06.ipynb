```python
#QUESTÃO 02 – Soma dos Números (2 pontos)
#---------------------------------------------------------

#Solicite ao usuário um número inteiro positivo.

#Utilizando um laço de repetição, calcule e exiba a soma de todos os números de 1 até o número informado.

n = int (input("Digite um número: "))
soma = 0

for i in range (n+1):
  soma += i

print (f"A soma de todos os números de 1 até {n} é {soma}.")
```

```python
#QUESTÃO 04 – Fatorial (4 pontos)
#---------------------------------------------------------

#Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.


n = int (input("Digite um número que você deseja saber o fatorial: "))
while n < 0:
  if n < 0:
    print ("Erro, Entrada Inválida! Digite uma entrada positiva! ")

  n = int (input("Digite um número que você deseja saber o fatorial: "))


for i in reversed(range (1, n)):
  n *= i

print (n)
```

```python
#QUESTÃO 05 – Caixa Eletrônico (5 pontos)
#---------------------------------------------------------

#Um caixa eletrônico possui notas de:

#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1

#Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.

#Exemplo:

#Digite o valor do saque: 287

#Saída esperada:

#2 nota(s) de R$100
#1 nota(s) de R$50
#1 nota(s) de R$20
#1 nota(s) de R$10
#1 nota(s) de R$5
#1 nota(s) de R$2


cem = 0
cinquenta = 0
vinte = 0
dez = 0
cinco = 0
dois = 0
um = 0

saque = int (input("Digite o valor desejado para saque: "))

while saque > 0:

  if saque >= 100:
    saque -= 100
    cem += 1

  elif saque >= 50:
    saque -= 50
    cinquenta += 1

  elif saque >= 20:
    saque -= 20
    vinte += 1

  elif saque >= 10:
    saque -= 10
    dez += 1

  elif saque >= 5:
    saque -= 5
    cinco += 1

  elif saque >= 2:
    saque -= 2
    dois += 1

  elif saque >= 1:
    saque -=1
    um += 1

if cem > 0:
    print (f"{cem} nota(s) de R$ 100.")

if cinquenta > 0:
    print (f"{cinquenta} nota(s) de R$ 50.")

if vinte > 0:
    print (f"{vinte} nota(s) de R$ 20.")

if dez > 0:
    print (f"{dez} nota(s) de R$ 10.")

if cinco > 0:
    print (f"{cinco} nota(s) de R$ 5.")

if dois > 0:
    print (f"{dois} nota(s) de R$ 2.")

if um > 0:
    print (f"{um} nota(s) de R$ 1.")
```
