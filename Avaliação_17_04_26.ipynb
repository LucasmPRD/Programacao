# 1 PAR OU ÍMPAR
FAZER CÓDIGO INFORMANDO SE O CÓDIGO É PAR OU ÍMPAR

```python
print ("-=-=-=PAR OU ÍMPAR?=-=-=-=")

n1 = int (input("Digite o número:"))
print ("-=-"*10)

if n1 % 2 == 0:
  print (f"{n1} é um número par!")

else:
  print (f"{n1} é um número ímpar!")
```

# 2. Número maior, menor ou igual?
Solicite dois números e mostre qual é o maior (ou se são iguais).

```python
print ("-=-=-=NÚMERO MAIOR OU MENOR?-=-=-=")

n1 = float (input ("Digite o primeiro número:"))
n2 = float (input ("Digite o segundo número:"))
print ("-=-" *10)

if n1 > n2:
  print (f"{n1} é maior que {n2}!")

elif n1 < n2:
  print (f"{n1} é menor que {n2}!")

else:
  print (f"{n1} é igual à {n2}!")

```

# 3. APROVAÇÃO DO ALUNO
Peça a nota de um aluno e informe:
o Aprovado (nota ≥ 7)
o Recuperação (5 ≤ nota &lt; 7)
o Reprovado (nota &lt; 5)

```python
print ("-=-=-=-APROVAÇÃO DO ALUNO-=-=-=-")

nome = str (input ("Digite o nome do aluno:"))
print ("-=-"*10)
print ("Digite apenas notas entre 0 e 10!")

n1 = float (input ("Digite a primeira nota:"))
n2 = float (input ("Digite a segunda nota:"))
n3 = float (input ("Digite a terceira nota:"))
n4 = float (input ("Digite a quarta nota:"))

media = (n1 + n2 + n3 + n4) / 4

print ("-=-"*10)

if media >= 7:
  print (f"{nome} foi Aprovado com a nota {media}!")

elif 5 <= media < 7:
  print (f"{nome} está de Recuperação com a nota {media}!")

else:
  print (f"{nome} foi Reprovado com a nota {media}!")
```

# 4. Classificação de idade
Receba a idade e classifique:
o Criança (0–12)
o Adolescente (13–17)
o Adulto (18–59)
o Idoso (60+)

```python
print ("-=-=-Classificador Etário-=-=-")

idade = int (input ("Digite a sua Idade:"))

print ("-=-"*10)

if 0 <= idade <= 12:
  print (f"Você tem {idade} anos, ou seja, você é Criança!")

elif 13 <= idade <= 17:
   print (f"Você tem {idade} anos, ou seja, você é Adolescente!")

elif 18 <= idade <= 59:
   print (f"Você tem {idade} anos, ou seja, você é Adulto!")

else:
   print (f"Você tem {idade} anos, ou seja, você é Idoso!")
```

#6. Positivo, negativo ou zero
Peça um número e diga se ele é positivo, negativo ou zero.

```python
print ("-=-=-Positivo, negativo ou zero-=-=-")

n = int (input ("Digite um número:"))
print ("-=-"*10)

if n == 0:
  print (f"O número escolhido é o zero!")

elif n >= 1:
  print ("O número escolhido é poitivo!")

else:
   print ("O número escolhido é negativo!")
```

#7. Calculadora simples
Peça dois números e uma operação (+, -, *, /). Use if/elif para calcular o
resultado.

```python
print ("-=-=-Calculadora-=-=-")
n1 = float (input ("Digite o primeiro número:"))
n2 = float (input ("Digite o segundo número:"))
print ("-=-"*10)

print ("Opção 1: ADIÇÃO")
print ("Opção 2: SUBTRAÇÃO")
print ("Opção 3: MULTIPLICAÇÃO")
print ("Opção 4: DIVISÃO")

o = int (input ("Digite o número da opção desejada:"))
print ("-=-"*10)
if o == 1:
  print (f" {n1 + n2}")

elif o == 2:
  print (f"{n1 - n2}")

elif o == 3:
  print (f"{n1 * n2}")

elif o == 4:
  print (f"{n1 / n2}")

else:
  print ("Digite uma Opção Válida!")
```

#10. Desconto em compras
Peça o valor de uma compra:
 ≥ 100: 10% de desconto
 ≥ 50 e &lt; 100: 5% de desconto
 &lt; 50: sem desconto

```python
c = float (input ("Digite o valor da compra:"))
print ("-=-")

if c >= 100:
  print (f"Você tem direto a 10 % de desconto! Sua compra ficou {c * 0.90}")

elif 100 > c >= 50:
  print (f"Você tem direto a 5 % de desconto! Sua compra ficou {c * 0.95}")

else:
  print (f"Você não tem direito de desconto! Sua compra ficou {c}")
```
