# Exercício 1 — Contagem

Faça um programa que peça um número ao usuário e mostre todos os números de 1 até esse número usando while.

```python
contador = 0

while contador != 11:
  print (contador)
  contador += 1
```

Exercício 2 — Contagem regressiva

Peça um número ao usuário e faça uma contagem até 0, mostrando cada número na tela.

```python
n = int (input('Digite um número:'))

while n != -1:
  print (n)
  n -= 1
```

#Exercício 4 — Soma dos números

Peça um número ao usuário e calcule a soma de todos os números de 1 até esse número.

```python
n = int (input('Digite um número:'))
soma = 0
while n != 0:
  soma += n
  n -= 1

print (f"A soma dos números é {soma}")
```

#Exercício 5 — Tabuada

Peça um número ao usuário e mostre a tabuada desse número de 1 até 10 usando while.

```python
contador = 0
n = int (input('Digite um número:'))

while contador != 11:
  print (f"{n} x {contador} = {n*contador}")
  contador += 1
```

#Exercício 6 — Senha

Crie um programa que peça uma senha ao usuário.

Enquanto a senha estiver incorreta, mostre:

Senha incorreta!

Quando o usuário digitar a senha correta, mostre:

Acesso permitido!

Considere que a senha correta seja "1234".

```python
sen = 0

while sen != "1234":
  sen = str (input("Digite sua senha: "))
  if sen != "1234":
    print ("Acesso Negado!")

  else:
    print ("Acesso permitido!")
```

Faça um programa que peça vários números ao usuário.

O programa deve continuar pedindo números até que o usuário digite 0.

Ao final, mostre a soma de todos os números digitados.

# Soma: 23

#Exercício 8 — Maior número

Peça números ao usuário até que ele digite 0.

Ao final, mostre qual foi o maior número digitado.

Exemplo:

Digite um número: 15

Digite um número: 8

Digite um número: 32

Digite um número: 20

Digite um número: 0


Maior número: 32

```python
maior = None

while True:
    numero = int(input("Digite um número ou 0 para parar671: "))

    if numero == 0:
        break

    if maior is None or numero > maior:
        maior = numero

print(f"O maior número é: {maior}")
```

#Exercício 9 — Média das notas

Faça um programa que peça 5 notas de um aluno usando while.

Ao final, calcule e mostre a média das notas.

Exemplo:

Digite a nota: 7

Digite a nota: 8

Digite a nota: 6

Digite a nota: 9

Digite a nota: 10


Média: 8.0

```python
notas = 1
media = 0
while True:
  n = int (input(f"Digite a {notas}º nota do aluno:"))
  media += n
  notas += 1
  if notas == 6:
    break

print (f"A média do aluno foi: {media / 5}")


```

#Exercício 10 — Caixa eletrônico

Faça um programa que comece com um saldo de R$ 1000,00.

Mostre um menu:

1 - Depositar

2 - Sacar

3 - Ver saldo

4 - Sair

```python
saldo = 1000


while True:
  print ("")
  print ("")
  print ("-=-=-=-= MENU =-=-=-=-")
  print ("")
  print ("1 - DEPOSITAR")
  print ("-=" * 10)
  print ("2 - SACAR")
  print ("-=" * 10)
  print ("3 - VER SALDO")
  print ("-="*10)
  print ("4 - SAIR")
  print ("")
  print ("")

  op = int (input("Digite o número da opção que deseja acessar do menu: "))

  if op == 1:
    dep = float (input("Digite o valor que deseja depositar: "))
    saldo += dep
    print ("Depósito realizado com sucesso!")

  elif op == 2:
    sac = float (input("Digite o valor que deseja sacar: "))
    if sac <= saldo:
      saldo -= sac
      print ("Saque realizado com sucesso!")

    else:
      print ("Saldo insuficiente... Saque um valor menor!")

  elif op == 3:
    print (f"Saldo atual: R${saldo}.")

  elif op == 4:
    print ("Saindo... Obrigado por utilizar o caixa eletrônico! ")
    break

  else:
    print ("Opção Inválida! Tente Novamente.")




```
