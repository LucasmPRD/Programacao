9. Positivo ou negativo

Crie uma função chamada verificar_numero(numero) que receba um número e informe se ele é:

· positivo;

· negativo;

· zero.

```python
def verificar_numero(numero):
  if numero > 0:
    print ("O número é positivo!")

  elif numero < 0:
    print ("O número é negativo!")

  else:
    print ("O número é 0!")

verificar_numero(0)
verificar_numero(-10)
verificar_numero(15)
```
