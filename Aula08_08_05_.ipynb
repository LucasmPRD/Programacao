```python
#1.Escrever a sequencia do 0 ao número

n = int (input("Digite o número desejado:"))


for i in range (0,n+1):
  print (i)
```

```python
# 2.Escrever ao contrário do número ao zero

n = int (input("Digite o número desejado:"))


for i in range (0, n+1):
    print ( n - i )
```

```python
#Esrever os números pares da variável de índice

n = int (input("Digite o número desejado:"))


for i in range (0,n+1):
    if i % 2 == 0:
      print (i)
```

```python
#4.Escrever os numeros impares da variável de índice

n = int (input("Digite o número desejado:"))

for i in range (0,n):
    if i % 2 != 0:
      print (i)
```

```python
#5. Soma das variaveis

n = int (input("Digite o número desejado:"))
soma = 0

for i in range (0,n):
    soma +=i
    print (f"o resultado da soma das variáveis é : {soma}")
```

```python
#6.Tabuada das variáveis

n = int (input("Digite o número desejado:"))


for i in range (0,11):
    multiplicacao = n * i
    print (f"{n} x {i} = {multiplicacao}")
```

```python
#7.Escrever uma palavra e contá-la

palavra = str (input("Digite a palavra desejada:"))

for i in (palavra):
  print (i)
```

```python

#8.Contar as vogais da palavra
palavra = str (input ("Digite a palavra desejada:"))

vogais= 0

for i in palavra.lower():
  if i in "aeiou":
    vogais += 1

print(f"A quantidade de vogais na palavra {palavra} é de: {vogais}")

```

```python
# 9. Maior número digitado, o usuário digita 4 números e o programa imprimi o maior

print ("Digite um número por campo! Haverão 4 campos de digitação!")
n = int (input("Digite um número: "))
n01 = int (input("Digite um número: "))
n02 = int (input("Digite um número: "))
n03 = int (input("Digite um número: "))

maior = max(n, n01, n02, n03)

print (f"O maior número digitado foi o: {maior}")
```

```python
# 10. O usuário digita um número entre 1 e 100 e o programa mostra quais são divisíveis por 3 e 5

n = int (input( "Digite um número inteiro entre 1 e 100:"))

for i in range (0, n):
  if i % 3 == 0 and i % 5 == 0:
    print (i)

```
