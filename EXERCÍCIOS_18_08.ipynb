#Exercício 1 – Idade para dirigir
Receba a idade de uma pessoa e diga se ela pode tirar carteira de motorista. A pessoa

#Exercício 2 – Par ou Ímpar
Receba um número e diga se ele é par ou ímpar.

```python
n = int (input("Digite um número e descubra se ele é par ou ímpar: "))

if n % 2 == 0:
  print (f"{n} é um número par!")

else:
  print (f"{n} é um número ímpar!")
```

#Exercício 3 – Maior de dois números
Receba dois números e mostre qual é o maior. Se os dois números forem iguais, diga que
eles são iguais.

```python
n1 = int (input("Digite o primeiro número: "))
n2 = int (input("Digite o segundo número: "))


if n1 > n2:
  print (f"{n1} é maior que {n2}.")

elif n1 == n2:
  print ("Eles são iguais.")

else:
  print (f"{n1} é menor que {n2}.")
```

#Exercício 7 – Classificação de idade
Receba a idade de uma pessoa e classifique-a de acordo com sua faixa etária:
 Menor de 12 anos → Criança
 De 12 a 17 anos → Adolescente
 De 18 a 59 anos → Adulto
 60 anos ou mais → Idoso

```python
idade = int (input("Digite sua idade: "))

if idade < 12:
  print ("Você é uma Criança!")

elif 12 <= idade <= 17:
  print ("Você é um Adolescente!")

elif 18 <= idade <= 59:
  print ("Você é um Adulto!")

else:
  print ("Você é Idoso!")
```

#Exercício 10 – Calculadora de IMC
Receba o peso e a altura de uma pessoa, calcule o IMC usando a fórmula:
IMC = peso / (altura × altura)
Depois, classifique o resultado:
 IMC menor que 18.5 → Abaixo do peso
 IMC de 18.5 a 24.9 → Peso normal
 IMC de 25 a 29.9 → Sobrepeso
 IMC maior ou igual a 30 → Obesidade

```python
peso = float (input("Digite seu peso em Kg: "))
alt = float (input("Digite sua altura em metros: "))

imc = peso / (alt * alt)

if imc < 18.5:
  print ("Você está Abaixo do peso!")

elif 18.5 <= imc <= 24.9:
  print ("Você está com o peso Normal!")

elif 25 <= imc <= 29.9:
  print ("Você está Sobrepeso")

elif imc >= 30:
  print ("Você tem Obesidade!")
```
